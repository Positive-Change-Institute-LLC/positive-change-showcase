# Hydra subsystem demo

This notebook demonstrates how capital, risk tolerance, and signal intensity change the Hydra yield corridor and engine decision.

```python
from pprint import pprint

from ppci_sws_proof.api.main import simulate_hydra
```

```python
baseline = simulate_hydra(capital=150_000, risk_tolerance=0.45, signal_intensity=0.1)
pprint(baseline)
```

```python
signals = {
    'defensive_signal': simulate_hydra(150_000, 0.45, -0.7),
    'neutral_signal': simulate_hydra(150_000, 0.45, 0.0),
    'offensive_signal': simulate_hydra(150_000, 0.45, 0.8),
}
pprint(signals)
```

```python
yield_adjustments = {
    'lower_risk': simulate_hydra(500_000, 0.2, 0.3)['yield_corridor'],
    'mid_risk': simulate_hydra(500_000, 0.5, 0.3)['yield_corridor'],
    'higher_risk': simulate_hydra(500_000, 0.8, 0.3)['yield_corridor'],
}
pprint(yield_adjustments)
```
